```python
from datasets import load_dataset

data = load_dataset("lmms-lab/bagel-example")
```

```python
data = data["train"]
```

```python
data
```

```python
import json
import tqdm

data_list = []
for d in tqdm.tqdm(data):
    d["messages"][0]["content"][0]["text"] = json.loads(d["messages"][0]["content"][0]["text"])["caption"]
    data_list.append(d)
```

```python
data_list[0]
```

```python
import datasets

final_data = datasets.Dataset.from_list(data_list)
```
