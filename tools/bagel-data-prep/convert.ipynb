```python
from datasets import load_dataset

data = load_dataset("lmms-lab/bagel-example")
```

```python
data = data["train"]
```

```python
data
```

```python
import json
import tqdm

def gen():
    for d in tqdm.tqdm(data):
        d["messages"][0]["content"][0]["text"] = json.loads(d["messages"][0]["content"][0]["text"])["caption"]
        yield d
```

```python
import datasets

final_data = datasets.Dataset.from_generator(gen)
```

```python
final_data.push_to_hub("lmms-lab/bagel-example")
```
