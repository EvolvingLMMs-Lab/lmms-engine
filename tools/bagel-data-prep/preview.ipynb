```python
import datasets
```

```python
dataset = datasets.load_dataset("pufanyi/bagel-example", "t2i")
```

```python
data = dataset["train"]
```

```python
data[0]["captions"]
```

```python
from PIL import Image
import io

def load_image(b64_image: str) -> Image.Image:
    image = Image.open(io.BytesIO(b64_image))
    return image
```

```python
def convert_to_lmms_format(_id: str, d: dict) -> dict:
    return {
        "id": _id,
        "messages": [
            {
                "role": "user",
                "content": [
                    {
                        "type": "text",
                        "text": d["captions"]
                    }
                ]
            },
            {
                "role": "assistant",
                "content": [
                    {
                        "type": "image_col",
                        "image_col": "images[0]"
                    }
                ]
            }
        ],
        "images": [
            load_image(d["image"])
        ]
    }
```

```python
from tqdm import tqdm

final_data_list = []
for i, d in enumerate(tqdm(data, desc="Converting to LMMs format", total=len(data))):
    final_data_list.append(convert_to_lmms_format(str(i), d))
```

```python
final_data_list[1]
```

```python
final_data = datasets.Dataset.from_list(final_data_list)
```

```python
final_data.push_to_hub("lmms-lab/bagel-example", "t2i")
```
